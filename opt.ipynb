```python
import datetime as dt
from pathlib import Path
import numpy as np

import pandas as pd

date = dt.date.today()
# date_str = f"{date.year}{date.month}{date.day}"
date_str = "20261014"
a = .25

DATE = Path(date_str)

sym = "A"
opts = pd.read_csv(f"{date_str}/options_snapshot.csv")
unds = pd.read_csv(f"{date_str}/underlyings_snapshot.csv")
```

```python
filtered_opts = opts[opts["volume"] > 0]
vol_sum_opts = filtered_opts.groupby("underlying")["volume"].sum()
vol_sum_opts = vol_sum_opts.rename("volume_sum")
vol_sum_opts
```

```python
filtered_opts = filtered_opts.merge(vol_sum_opts, on="underlying", how="left")
filtered_opts["vol_factor"] = filtered_opts["volume"] / filtered_opts["volume_sum"]
```

```python
filtered_opts
```

```python
unds[["underlying", "price"]]
```

```python
filtered_opts = filtered_opts.merge(unds[["underlying", "price"]], on="underlying", how="left")
filtered_opts = filtered_opts.rename(columns={"price":"underlying_price"})
filtered_opts
```

```python
filtered_opts["dist"] = abs(filtered_opts["strike"] - filtered_opts["underlying_price"]) / filtered_opts["underlying_price"]
filtered_opts["dist_factor"] = np.where(filtered_opts["dist"] <= a, (filtered_opts["dist"] - a) ** 2 / a ** 2, 0)
filtered_opts
```

```python
filtered_opts.loc[filtered_opts["dist_factor"] == 0]
```

```python
filtered_opts = filtered_opts.drop("comp_iv", axis=1)
```

```python
# syms = unds["underlying"]
num = filtered_opts["vol_factor"] * filtered_opts["dist_factor"] * filtered_opts["iv"]
den = filtered_opts["vol_factor"] * filtered_opts["dist_factor"]
filtered_opts["num"] = num
filtered_opts["den"] = den
summed = filtered_opts.groupby("underlying", as_index=False)[["num", "den"]].sum()
summed["comp_iv"] = summed["num"] / summed["den"]
summed
```

```python
final_opts = filtered_opts.merge(summed[["underlying", "comp_iv"]], on="underlying", how="left")
final_opts
```

```python
unds_ivs = unds[["underlying", "iv30"]]
unds_ivs = unds_ivs.merge(summed[["underlying", "comp_iv"]], on="underlying", how="left")
unds_ivs
```

```python
unds_ivs["dif"] = unds_ivs["iv30"] / 100 - unds_ivs["comp_iv"]
unds_ivs
```

```python
min_dif, max_dif = min(unds_ivs["dif"]), max(unds_ivs["dif"])
print("min dif b/t composite iv and listed iv is...", min_dif)
print("max dif b/t composite iv and lsited iv is...", max_dif)
```
